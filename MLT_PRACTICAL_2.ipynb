```python
#here to implement simple linear regression
#Step 1:Importing Required Libraries
#Step 2:Read the Data
#Step 3:Model Selection and Fitting
#Step 4:Predictiung value using linear equation y=mx+c
#Step 5:Visualiazing the Data
```

**Step 1:Importing Required Libraries**

```python
#Step 1:Importing Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**Step 2:Read the Data**


```python
data={"hours":[1,2,3,4,5],
      "marks":[50,60,80,90,100]}
df=pd.DataFrame(data)
print(df)
```

```python
X=df[["hours"]].values
Y=df[["marks"]].values
```

**Model Fitting**

> Equation of linear regression is y=mx+c ,**what is slope** it is y/x,changes in dependent variable when their is unit change in independent variable.

**What is intercept**=even in the absence of x there will be some value i.e is even if,student will not studying he/she will getting the marks i.e that is basically intercept according to this example.

So basically here we are finding change in marks when this is unit change in no of hours studied.


```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X,Y)
```

```python
slope=model.coef_[0]
intercept=model.intercept_
```

```python
#"f-string" (formatted string literal )in python
#.2f this is format specifier used within curly braces of an f-string
print(f"slope(m): {slope}")
print(f"intercept (b): {intercept}")
print(f"Equation of regression line: y={slope[0]:.2f}  x + {intercept[0]:.2f}")
```

```python
df["Predicted Marks"]=model.predict(X)
```

```python
#here keep in mind the data which we have provided to the model is also linear
df.head()
```

**Visualizing**

```python
plt.scatter(df['hours'],df['marks'],color='blue',label='Actual marks')
```

**Plotting the equation**

```python
plt.plot(df['hours'],df['Predicted Marks'],color='red',label="Regession Line")
plt.xlabel('Hours Studied')
plt.ylabel('Marks')
plt.title('Linear Regression:hours vs marks')
plt.legend()
plt.show()
```
