### Step 3 :Training the model

defining Independent features and dependent features

iloc is an attribute used for integer-location based indexing in pandas DataFrames. For example:

df.iloc[0] selects the first row.

df.iloc[:, 0] selects the first column.

df.iloc[0:5, 0:2] selects the first 5 rows and the first 2 columns

```python
X= df.iloc[:,:-1] # This selects all rows and all columns except the last one of the DataFrame `df`.
y=iris.target # This assigns the target variable from the iris dataset to the variable `y`.
```

```python
X
```

### Step 4: splitting the data into training and testing datasets

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
X_train
```

Implementing decision tree

!!! creating our model

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
tree_model=DecisionTreeClassifier()
```

```python
#The fit function is a standard method in scikit-learn for training a model.It is used to train the model.
tree_model.fit(X_train,y_train)
```

### step 6 :Testing our model using testing data

```python
X_test.head()
```

```python
predictions = tree_model.predict(X_test)
```

```python
print("Predictions:")
print(predictions)
```

plotting our decision tree

```python
from sklearn import tree
plt.figure(figsize=(12,8))
tree.plot_tree(tree_model,filled=True)
#here by default gini is used, Gini, specifically Gini Impurity or Gini Index, is a metric used to evaluate the quality
# of a split in a node.
#The Gini Impurity ranges from 0 to 1.
# 0: Represents perfect purity, where all instances in the node belong to the same class
```
